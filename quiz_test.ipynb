```python
from atw.data_loader import load_countries
from atw.questions import generate_quiz
from atw.scoring import check_answer, calculate_results

countries = load_countries("data/countries.json")
countries_list = countries.to_dict("records")

quiz = generate_quiz(countries_list, 10)
```

```python
question = quiz[0]

result = check_answer(
    question,
    question["correct_answer"],
    "Medium"
)

print(result)
```

```python
assert result["is_correct"] == True
assert result["points"] == 2

print("Scoring test passed")
```

```python
results = []

for i, question in enumerate(quiz, 1):
    print(f"\nQuestion {i}")
    print(question["question"])

    for number, option in enumerate(question["options"], 1):
        print(f"{number}. {option}")

    while True:
        answer = input("Your answer (1-4): ").strip()

        if answer in ["1", "2", "3", "4"]:
            break

        print("Please enter a number from 1 to 4.")

    selected_answer = question["options"][int(answer) - 1]

    result = check_answer(question, selected_answer, "Easy")
    results.append(result)

    if result["is_correct"]:
        print("Correct!")
    else:
        print("Incorrect!")
        print("Correct answer:", question["correct_answer"])
```

```python
summary = calculate_results(results)

print("Quiz completed!")
print("Correct answers:", summary["correct_answers"])
print("Total questions:", summary["total_questions"])
print("Score:", summary["total_points"])
print("Maximum score:", summary["max_points"])
print("Accuracy:", summary["accuracy"], "%")
```
